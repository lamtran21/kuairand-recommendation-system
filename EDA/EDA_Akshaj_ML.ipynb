```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
# Convert duration from milliseconds to seconds for better readability
sns.histplot(video_features_basic['video_duration'] / 1000, bins=50, color='skyblue', kde=True)
plt.title('Distribution of Video Duration (Seconds)')
plt.xlabel('Duration (Seconds)')
plt.ylabel('Number of Videos')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=video_features_basic, x='upload_type', 
              order=video_features_basic['upload_type'].value_counts().index, 
              palette='Set3')
plt.title('Distribution of Upload Types')
plt.xlabel('Upload Type')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45) # Rotate for better label readability
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=video_features_basic, x='music_type', 
              order=video_features_basic['music_type'].value_counts().index, 
              palette='pastel')
plt.title('Distribution of Music Types')
plt.xlabel('Music Type ID')
plt.ylabel('Number of Videos')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Explode the comma-separated tags into individual items
tags_expanded = video_features_basic['tag'].astype(str).str.split(',', expand=True).stack()
top_tags = tags_expanded.value_counts().head(10)

sns.barplot(x=top_tags.index, y=top_tags.values, palette='viridis')
plt.title('Top 10 Most Frequent Video Tags')
plt.xlabel('Tag ID')
plt.ylabel('Number of Videos')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=video_features_basic, x='server_width', y='server_height', 
                alpha=0.5, color='crimson')
plt.title('Video Resolutions (Width vs. Height)')
plt.xlabel('Server Width (Pixels)')
plt.ylabel('Server Height (Pixels)')

# Add a reference line for square videos (1:1 ratio)
plt.plot([0, 2000], [0, 2000], 'k--', lw=1, label='1:1 Ratio')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
# Using a log scale because play counts usually have a massive variance
sns.histplot(video_features_statistics['play_cnt'] + 1, bins=50, log_scale=True, color='dodgerblue', kde=False)
plt.title('Distribution of Video Play Counts (Log Scale)')
plt.xlabel('Total Plays per Video (Log Scale)')
plt.ylabel('Frequency (Number of Videos)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=video_features_statistics, x='show_cnt', y='play_cnt', 
                alpha=0.5, color='darkorange')
plt.title('Video Engagement Funnel: Shows vs. Plays')
plt.xlabel('Show Count (Impressions)')
plt.ylabel('Play Count (Actual Views)')

# Add a reference line for a 100% play-through rate
max_val = min(video_features_statistics['show_cnt'].max(), video_features_statistics['play_cnt'].max())
plt.plot([0, max_val], [0, max_val], 'k--', lw=1, label='100% Play Rate')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Calculate completion rate per video (completed plays / total plays)
# Fill NaNs with 0 in case of division by zero
video_features_statistics['completion_rate'] = (video_features_statistics['complete_play_cnt'] / video_features_statistics['play_cnt']).fillna(0)

sns.histplot(video_features_statistics['completion_rate'], bins=40, color='mediumseagreen', kde=True)
plt.title('Distribution of Average Video Completion Rates')
plt.xlabel('Completion Rate (Complete Plays / Total Plays)')
plt.ylabel('Number of Videos')
plt.show()
```

```python
plt.figure(figsize=(10, 8))

# Select key interaction metrics for the heatmap
engagement_cols = [
    'play_cnt', 'like_cnt', 'comment_cnt', 'share_cnt', 
    'follow_cnt', 'collect_cnt', 'download_cnt'
]

# Calculate the correlation matrix
corr_matrix = video_features_statistics[engagement_cols].corr()

# Plot the heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5, vmin=0, vmax=1)
plt.title('Correlation Heatmap of Key Engagement Metrics')
plt.xticks(rotation=45, ha='right')
plt.show()
```
